## Import modules

```python
from src.ingestion.loaders import MDNLoader, DataLoader
from src.config import Config
```

```python
# Add repo root to Python path so we can import from src
import sys
from pathlib import Path

repo_root = Path.cwd().parent
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

print(f"✅ Added to path: {repo_root}")
```

## Initialize MDN loader

```python
# Initialize loader with default settings
loader = MDNLoader()

print(f"Repository: {loader.repo_url}")
print(f"Local directory: {loader.repo_dir}")
print(f"Included subdirectories:")
for subdir in loader.include_subdirs:
    print(f"  - {subdir.relative_to(loader.repo_dir)}")
```

## Load raw documents (with frontmatter)

```python
# Load raw docs and save to JSONL
raw_docs = loader.load_raw(
    output_path=Config.MVP_KB_OUTPUTS / "mdn_learn_docs_raw.jsonl"
)

print(f"\n✅ Loaded {len(raw_docs)} raw documents")
print(f"\nSample document:")
print(f"  Title: {raw_docs[0]['title']}")
print(f"  Doc ID: {raw_docs[0]['doc_id']}")
print(f"  URL: {raw_docs[0]['source_url_guess']}")
```

## Load cleaned documents (ready for embedding)

```python
# Load clean docs (removes macros, HTML, normalizes whitespace)
clean_docs = loader.load_clean(
    output_path=Config.MVP_KB_OUTPUTS / "mdn_learn_docs_clean.jsonl"
)

print(f"\n✅ Loaded {len(clean_docs)} cleaned documents")
print(f"\nFirst document preview:")
print(f"  Title: {clean_docs[0]['title']}")
print(f"  Text length: {len(clean_docs[0]['text'])} chars")
print(f"\nFirst 300 chars:")
print(clean_docs[0]['text'][:300])
```

## Load existing JSONL files (if already generated)

```python
# Alternative: Load from existing JSONL file without re-processing
from pathlib import Path

clean_jsonl_path = Config.MVP_KB_OUTPUTS / "mdn_learn_docs_clean.jsonl"

if clean_jsonl_path.exists():
    docs = DataLoader.load_jsonl(clean_jsonl_path)
    print(f"📄 Loaded {len(docs)} documents from existing file")
else:
    print("ℹ️ File doesn't exist yet. Run cells above to generate it.")
```
